## Python EDA

```python
# 1 How many jobs are there by country?

df["country_name"].value_counts()
```

```python
# which job categories have the most listings.

df["category"].value_counts()
```

```python
df["is_ai_related"].value_counts()
```

```python
# AI-related jobs by country. This will be more useful than just looking at the overall AI count.
df.groupby(["country_name", "is_ai_related"]).size()
```

```python
# analyze jobs by year to understand how the job market changes over time.
df.groupby(df["created_date"].dt.year).size()
```

```python
# jobs by month in 2026. This will be much more meaningful because most of our data is from 2026.
df[df["created_date"].dt.year == 2026]["created_date"].dt.month.value_counts().sort_index()
```

```python
# which companies have the most job listings.
df["company"].value_counts().head(10)
```

```python
# contract type — permanent vs contract vs unknown.
df["contract_type"].value_counts()
```

```python
df["contract_time"].value_counts()
```

```python
# find the average salary by country.
df.groupby("country_name")["salary_avg"].mean()
```

```python
df.groupby(["country_name", "currency"])["salary_avg"].mean()
```

```python
# Salary range by country
df.groupby("country_name")["salary_avg"].agg(["count", "min", "max", "mean", "median"])
```

```python
# identify the Indian salary records
df[df["country_name"] == "India"][
    ["title", "salary_min", "salary_max", "salary_avg"]
].sort_values("salary_avg", ascending=False)
```

```python
# whether AI-related jobs have higher salaries within each country.
df.groupby(["country_name", "is_ai_related"])["salary_avg"].agg(
    ["count", "mean", "median"]
)
```

```python
# check predicted vs non-predicted salaries.
df.groupby(["country_name", "salary_is_predicted"])["salary_avg"].agg(
    ["count", "mean", "median"]
)
```

```python
# Job Title Analysis
df["title"].value_counts().head(15)
```

```python
# which categories have the most AI-related jobs.
df[df["is_ai_related"] == True]["category"].value_counts().head(10)
```

```python
# where the jobs are concentrated within each country.

df.groupby(["country_name", "location"]).size().sort_values(ascending=False).head(15)
```

```python
# Which companies are posting AI-related jobs?
df[df["is_ai_related"] == True]["company"].value_counts().head(15)
```

```python
df[df["is_ai_related"] == True].groupby(
    ["country_name", "company"]
).size().sort_values(ascending=False).head(15)
```

```python
# Job title + AI relationship
# ind the most common AI-related job titles specifically.

df[df["is_ai_related"] == True]["title"].value_counts().head(15)
```

```python
# AI percentage by country
df.groupby("country_name")["is_ai_related"].mean().mul(100).round(2)
```
